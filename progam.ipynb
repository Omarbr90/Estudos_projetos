```python
import itertools
from collections import Counter
import numpy as np

def solicitar_jogos():
    jogos = []
    for i in range(2):  # Solicita dois jogos ao usuário
        while True:
            try:
                jogo = list(map(int, input(f"Digite os 15 números do jogo {i + 1} (separados por espaço): ").split()))
                if len(jogo) != 15 or any(n < 1 or n > 25 for n in jogo):
                    raise ValueError("Você deve digitar exatamente 15 números entre 1 e 25.")
                jogos.append(jogo)
                break
            except ValueError as e:
                print(e)
    return jogos

def calcular_frequencia_numeros(jogos):
    todos_numeros = [num for jogo in jogos for num in jogo]
    frequencia = Counter(todos_numeros)
    return frequencia

def gerar_combinacoes_mais_frequentes(frequencia):
    numeros_frequentes = [num for num, freq in frequencia.most_common(20)]
    combinacoes = list(itertools.combinations(numeros_frequentes, 15))
    return combinacoes

def filtrar_combinacoes(combinacoes, frequencia):
    combinacoes_filtradas = []
    
    for combinacao in combinacoes:
        score_frequencia = sum(frequencia[num] * 2 for num in combinacao)
        penalizacao_consecutivos = sum(1 for j in range(len(combinacao) - 1) if combinacao[j] + 1 == combinacao[j + 1])
        score_total = score_frequencia - penalizacao_consecutivos
        combinacoes_filtradas.append((combinacao, score_total))
    
    combinacoes_filtradas.sort(key=lambda x: x[1], reverse=True)
    return combinacoes_filtradas[:10]

def main():
    print("Bem-vindo ao gerador de possíveis jogos!")
    
    jogos = solicitar_jogos()
    frequencia = calcular_frequencia_numeros(jogos)
    combinacoes = gerar_combinacoes_mais_frequentes(frequencia)
    melhores_jogos = filtrar_combinacoes(combinacoes, frequencia)
    
    print("\nPossíveis jogos baseados na análise dos dois últimos sorteios:")
    for i, (jogo, score) in enumerate(melhores_jogos, 1):
        jogo_ordenado = sorted(jogo)  # Ordena o jogo em ordem crescente
        print(f"Jogo {i}: {jogo_ordenado} (Score: {score})")

if __name__ == "__main__":
    main()
```
